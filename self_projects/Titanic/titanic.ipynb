```python
df_train=df_train.drop('Cabin',axis=1)
df_test=df_test.drop('Cabin',axis=1)
```

```python
for col in df_train.columns:
        if df_train[col].dtype=='object':
            df_train[col].fillna(df_train[col].mode()[0],inplace=True)
        else:
            df_train[col].fillna(df_train[col].mean(),inplace=True)
```

```python
for col in df_test.columns:
        if df_test[col].dtype=='object':
            df_test[col].fillna(df_test[col].mode()[0],inplace=True)
        else:
            df_test[col].fillna(df_test[col].mean(),inplace=True)
```

```python
from sklearn.preprocessing import LabelEncoder
label=LabelEncoder()
cata=df_train.select_dtypes(include='object').columns
for col in cata:
    df_train[col]=label.fit_transform(df_train[col])
```

```python
from sklearn.preprocessing import StandardScaler
scale=StandardScaler()
X=scale.fit_transform(X)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_val,y_train,y_val=train_test_split(X,y,test_size=0.2,stratify=y)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from xgboost import XGBClassifier
models={
        'SVM':SVC(),
    'knn':KNeighborsClassifier(),
    'logstic regression':LogisticRegression(),
    'ID3':DecisionTreeClassifier(),
    'random forest':RandomForestClassifier(),
    'XG_boost':XGBClassifier()

} 
```

```python
from sklearn.model_selection import cross_val_score
from datetime import datetime

score=[]
for model_name,model_per in models.items():
    start_time = datetime.now()
    test=cross_val_score(model_per,X,y,cv=5,scoring='accuracy')
    score.append({'model name': model_name, 'score': test.mean()})
    # time_consumed = datetime.now() - start_time
    # print("Time Consumed for :",model_name, time_consumed)
score=pd.DataFrame(score)
score
```

```python
from sklearn.model_selection import GridSearchCV

mod = RandomForestClassifier(random_state=42)

param_grid = {
    'n_estimators': [100, 200, 500],
    'max_depth': [None, 10, 20, 30],
    'min_samples_split': [2, 5, 10],
}
grid_search=GridSearchCV(
    estimator=mod,
    param_grid=param_grid,
    cv=5,
    n_jobs=-1,
    scoring='accuracy'
)
grid_search.fit(X_train,y_train)
print("Best parameters:", grid_search.best_params_)
print("Best score:",grid_search.best_score_)
```

```python
from sklearn.metrics import accuracy_score,classification_report,ConfusionMatrixDisplay
from sklearn.ensemble import RandomForestClassifier
mod=RandomForestClassifier(random_state=42)
# grid_search.best_estimator_
mod.fit(X_train,y_train)
y_pred=mod.predict(X_val)
accuracy= accuracy_score(y_val,y_pred)
```

```python
print('accuracy',accuracy)
print ('classifcation report',classification_report(y_val,y_pred))
print('confusion matrix',ConfusionMatrixDisplay.from_predictions(y_val,y_pred))
```
